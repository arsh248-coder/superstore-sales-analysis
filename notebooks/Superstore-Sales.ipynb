```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("superstore_sales.csv", encoding="latin1")
print(df.size)
```

```python
print(df.columns)
print(type(df['Order Date']))
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
print(df["Order Date"])
```

```python
df.fillna(value="Empty", inplace=True)
print(df)
```

```python
df.info()
```

```python
df.describe()
```

```python
for i in df.columns:
    print(df[i].value_counts())
```

```python
duplicate_rows = df[df.duplicated()]
print(duplicate_rows)
```

```python
# Aggregate per product
product_summary = df.groupby(['Product Name', 'Category', 'Sub-Category']).agg({
    'Sales': 'sum',
    'Quantity': 'sum',
    'Discount': 'mean',   # avg discount per product
    'Profit': 'sum'
}).reset_index()

# --- Filter for products with BOTH high sales and low discount ---

# Step 1: Define thresholds (median or quantile is safer than absolute min/max)
high_sales_threshold = product_summary['Sales'].quantile(0.75)   # top 25% sales
low_discount_threshold = product_summary['Discount'].quantile(0.25)  # bottom 25% discounts

# Step 2: Filter
filtered_products = product_summary[
    (product_summary['Sales'] >= high_sales_threshold) &
    (product_summary['Discount'] <= low_discount_threshold)
]

# Step 3: From filtered products, take the top 10 by Sales
top10_filtered = filtered_products.nlargest(10, 'Sales')

print(top10_filtered)

# Save to CSV
top10_filtered.to_csv("top10_high_sales_low_discount.csv", index=False)

# --- Plot them ---
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12,6))
sns.barplot(x='Sales', y='Product Name', data=top10_filtered, palette="viridis")
plt.title("Top 10 Products with High Sales and Low Discount")
plt.xlabel("Total Sales")
plt.ylabel("Product Name")
plt.show()
```

```python
# Extract Year-Month
df['YearMonth'] = df['Order Date'].dt.to_period('M')

# Aggregate sales by month
monthly_sales = df.groupby('YearMonth')['Sales'].sum().reset_index()
monthly_sales['YearMonth'] = monthly_sales['YearMonth'].dt.to_timestamp()

print(monthly_sales)
print(monthly_sales.max())
print(monthly_sales.min())

import matplotlib.pyplot as plt
import seaborn as sns

# Monthly sales (already aggregated)
plt.figure(figsize=(10,5))
sns.lineplot(x='YearMonth', y='Sales', data=monthly_sales, marker='o')
plt.xticks(rotation=45)
plt.title("Monthly Sales Trend")
plt.ylabel("Total Sales")
plt.xlabel("Month")
plt.show()
monthly_sales.to_csv("monthly_sales.csv", index=False)
```

```python
#Extract yearly average sales
df['Year'] = df['Order Date'].dt.year
avg_sales_per_year = df.groupby('Year')['Sales'].mean().reset_index()
print(avg_sales_per_year)
plt.figure(figsize=(8,5))
sns.barplot(x='Year', y='Sales', data=avg_sales_per_year, palette="viridis")
plt.title("Average Sales per Year")
plt.ylabel("Average Sales")
plt.show()
avg_sales_per_year.to_csv("avg_sales_per_year.csv", index=False)
```

```python
# Extract year from Order Date
df['Year'] = df['Order Date'].dt.year

# Group by Year and Region to get average sales
avg_sales_per_year_region = df.groupby(['Year', 'Region'])['Sales'].mean().reset_index()

print(avg_sales_per_year_region)

# Plot
plt.figure(figsize=(10,6))
sns.barplot(x='Year', y='Sales', hue='Region', data=avg_sales_per_year_region, palette="viridis")
plt.title("Average Sales per Year by Region")
plt.ylabel("Average Sales")
plt.xlabel("Year")
plt.legend(title="Region")
plt.show()

# Save to CSV
avg_sales_per_year_region.to_csv("avg_sales_per_year_region.csv", index=False)
```

```python
# First, create a Delivery Time column (difference between Ship Date and Order Date)
df['Delivery Time'] = (df['Ship Date'] - df['Order Date']).dt.days

# Group by Customer Name: total sales, total orders, and average delivery time
top_customers = df.groupby('Customer Name').agg({
    'Sales': 'sum',
    'Order ID': 'count',       # total number of orders
    'Delivery Time': 'mean'
}).reset_index()

# Rename for clarity
top_customers = top_customers.rename(columns={'Order ID': 'Total Orders'})

# Sort by Sales
top_customers = top_customers.sort_values(by='Sales', ascending=False)

# Keep only Top 10
top_10_customers = top_customers.head(10)

# Save to CSV
top_10_customers.to_csv("top_customers_with_delivery.csv", index=False)

print(top_10_customers)

# Plot Top 10 Customers by Total Orders
plt.figure(figsize=(12,5))
ax = sns.barplot(x='Customer Name', y='Total Orders', data=top_10_customers, palette="coolwarm")
plt.xticks(rotation=45)
plt.title("Top 10 Customers by Orders (with Avg Delivery Time)")
plt.ylabel("Total Orders")

# Add average delivery time as labels above each bar
for bar, delivery_time in zip(ax.patches, top_10_customers['Delivery Time']):
    x = bar.get_x() + bar.get_width() / 2   # center of the bar
    y = bar.get_height()                    # bar height
    ax.text(x, y + 0.5, f"{delivery_time:.1f} days", 
            ha='center', va='bottom', fontsize=9, color='black')

plt.show()

```

```python
# Bottom 10 customers by Total Orders
bottom_customers = top_customers.sort_values(by='Total Orders', ascending=True).head(10)

# Save to CSV
bottom_customers.to_csv("bottom_customers_with_delivery.csv", index=False)

print(bottom_customers)

plt.figure(figsize=(12,5))
ax = sns.barplot(x='Customer Name', y='Total Orders', data=bottom_customers, palette="coolwarm")
plt.xticks(rotation=60)
plt.title("Bottom 10 Customers by Orders (with Avg Delivery Time)")
plt.ylabel("Total Orders")

# Annotate with Delivery Time safely
for i, row in enumerate(bottom_customers.itertuples(index=False)):
    delivery_time = getattr(row, 'Delivery Time')  # Use actual column name
    total_orders = getattr(row, 'Total Orders')
    ax.text(i, total_orders + 0.02*total_orders, f"{delivery_time:.1f} days", 
            ha='center', va='bottom', fontsize=9, color='black')

plt.show()
```

```python
# Group by Region and Product, sum sales
product_region = df.groupby(['Region', 'Product Name']).agg({
    'Sales': 'sum'
}).reset_index()

# For each region, find the product with max sales
top_products_by_region = product_region.loc[
    product_region.groupby('Region')['Sales'].idxmax()
]

# Sort results for readability
top_products_by_region = top_products_by_region.sort_values(by='Sales', ascending=False)
print(top_products_by_region)
plt.figure(figsize=(12,6))
sns.barplot(x='Sales', y='Product Name', hue='Region', data=top_products_by_region, dodge=False)
plt.title("Top Product by Region")
plt.xlabel("Sales")
plt.ylabel("Product Name")
plt.show()
top_products_by_region.to_csv("top_products_by_region.csv", index=False)
```

```python
#Extracting the late deliveries
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])

# Delivery time in days
df['Delivery Days'] = (df['Ship Date'] - df['Order Date']).dt.days
df['Late Delivery'] = df['Delivery Days'] > 4
late_summary = df.groupby('Ship Mode')['Late Delivery'].mean().reset_index()
late_summary['Late Delivery %'] = late_summary['Late Delivery'] * 100
print(late_summary)
plt.figure(figsize=(6,4))
sns.barplot(x='Ship Mode', y='Late Delivery %', data=late_summary, palette="pastel")
plt.title("Late Delivery Percentage by Ship Mode")
plt.ylabel("Late Delivery (%)")
plt.show()
late_summary.to_csv("late_summary.csv", index=False)
```
